```python
import copy
goal = [
        [1, 3, 4],
        [8, 6, 2],
        [7, 0, 5]]
```

<b>Solving 8 Puzzel Problem Using A* (number of misplaced heuristic function)</b>

Example Action:
<pre>
2 8 3           1 2 3
1 6 4   ->      8 0 4
7 0 5           7 6 5
</pre>

Calculate heuristic Value

```python
# The heuristic function is the number of misplaced tiles (blank not included).
def h1(current, goal):
    sum = 0
    for x in range(3):
        for y in range(3):
            if current[x][y]!=goal[x][y] and goal[x][y]!=0:
                sum+=1
    return sum


test = [
        [7, 2, 4],
        [5, 0, 6],
        [8, 3, 1]]
test_goal = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8]]
h1(test,test_goal)
```

```python
def get_zero(state):
    for x in range(len(state)):
        for y in range(len(state[0])):
            if state[x][y]==0:
                return (x,y)

def get_neighboors(main):
    # Find 0 in matrix
    x,y = get_zero(main)
    
    neighboors = []
    # Top
    if x-1>=0:
        top = copy.deepcopy(main)
        top[x][y],top[x-1][y] = top[x-1][y],top[x][y]
        neighboors.append(top)

    # Right
    if y+1<=2:
        right = copy.deepcopy(main)
        right[x][y],right[x][y+1] = right[x][y+1],right[x][y]
        neighboors.append(right)

    # Down
    if x+1<=2:
        down = copy.deepcopy(main)
        down[x][y],down[x+1][y] = down[x+1][y],down[x][y]
        neighboors.append(down)

    # Left
    if y-1>=0:
        left = copy.deepcopy(main)
        left[x][y],left[x][y-1] = left[x][y-1],left[x][y]
        neighboors.append(left)

    return neighboors

def A_Star(start,goal):
    visited = []
    # The frontier is a priorty queue
    # fronter = [ ( actual path cost+ state heuristic cost , state, path ) ] 
    fronter = [(0+h1(start,goal),start,[start])]
    while len(fronter)>0:
        fronter.sort()
        # curr = ( actual path cost+ state heuristic cost , state, path )
        curr_cost, curr_state, curr_path = fronter.pop(0)
        curr_actual_path_cost = curr_cost-h1(curr_state,goal)
        if curr_state==goal:
            return curr_path
        if curr_state not in visited:
            visited.append(curr_state)
            neighboors = get_neighboors(curr_state)
            for child in neighboors:
                fronter.append(( curr_actual_path_cost+ 1 +h1(child,goal), child, curr_path+[child] ))
    return "No Solution found!"
```

Display State <b>(Optional)</b>

```python
def display_state(state):
    for row in range(len(state)):
        for col in range(len(state)):
            if state[row][col]!=0:
                print(state[row][col],end='\t')
            else:
                print(' ',end='\t')
        print()

def display_solution(solution):
    # Display actions
    for box in solution:
            display_state(box)
            print("\n")
```

Executions

```python
easy = [
        [1, 2, 3],
        [8, 0, 4],
        [7, 6, 5]]

solution,iterations = A_Star(easy,goal)
print(f"EASY Soultion: {len(solution)} Steps\nNumber of iterations = {iterations}")
display_solution(solution)
```
