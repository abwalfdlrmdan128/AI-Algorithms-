```python
start_state = [
    [7,2,4],
    [5,0,6],
    [8,3,1]
]
goal_state = [
    [0,1,2],
    [3,4,5],
    [6,7,8]
]
```

Functions

```python
'''
For example:


2 8 3           1 2 3
1 6 4   ->      8 0 4
7 0 5           7 6 5


'''

def get_zero(state):
    pos = [0,0]
    for i in range(len(state)):
        for j in range(len(state[0])):
            if state[i][j]==0:
                pos = [i,j]
                return pos
    

def move_up(state):
    # Find 0 in matrix
    x,y = get_zero(state)
    
    # Top
    if x-1>=0:
        state[x][y],state[x-1][y] = state[x-1][y],state[x][y]
        return state
    else:
        None

def move_right(state):
    x,y = get_zero(state)
    # Right
    if y+1<=2:
        state[x][y],state[x][y+1] = state[x][y+1],state[x][y]
        return state
    else:
        return None

def move_down(state):
    x,y = get_zero(state)
    # Down
    if x+1<=2:
        state[x][y],state[x+1][y] = state[x+1][y],state[x][y]
        return state
    else:
        return None

def move_left(state):
    x,y = get_zero(state)
    # Left
    if y-1>=0:
        state[x][y],state[x][y-1] = state[x][y-1],state[x][y]
        return state
    else:
        return None

def display_state(state):
    for row in range(len(state)):
        for col in range(len(state)):
            if state[row][col]!=0:
                print(state[row][col],end='\t')
            else:
                print(' ',end='\t')
        print()
```

Testing

```python
print("Goal State")
display_state(goal_state)
```

```python
current_state = start_state
while True:
    clear_output()
    # OR os.system('cls')
    display_state(current_state)
    action = input("Choose the action (top/right/down/left)")
    if action=='top':
        x = move_up(current_state)
        if x!=None:
            current_state = x

    elif action=='down':
        x = move_down(current_state)
        if x!=None:
            current_state = x

    elif action=='left':
        x = move_left(current_state)
        if x!=None:
            current_state = x
    
    elif action=='right':
        x = move_right(current_state)
        if x!=None:
            current_state = x
            
    else:
        break
```
