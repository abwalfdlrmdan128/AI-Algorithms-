8-puzzle Problem

![](https://miro.medium.com/v2/resize:fit:1358/1*yekmcvT48y6mB8dIcK967Q.png)

```python
import copy
goal = [
        [1, 3, 4],
        [8, 6, 2],
        [7, 0, 5]]
```

Solving 8-puzzle Problem using <b>Depth-first Search</b>

Example Action:
<pre>
2 8 3           1 2 3
1 6 4   ->      8 0 4
7 0 5           7 6 5
</pre>

```python
def get_zero(state):
    for x in range(len(state)):
        for y in range(len(state[0])):
            if state[x][y]==0:
                return (x,y)

def get_neighboors(main):
    # Find 0 in matrix
    x,y = get_zero(main)
            
    neighboors = []
    # Top
    if x-1>=0:
        top = copy.deepcopy(main)
        top[x][y],top[x-1][y] = top[x-1][y],top[x][y]
        neighboors.append(top)

    # Right
    if y+1<=2:
        right = copy.deepcopy(main)
        right[x][y],right[x][y+1] = right[x][y+1],right[x][y]
        neighboors.append(right)

    # Down
    if x+1<=2:
        down = copy.deepcopy(main)
        down[x][y],down[x+1][y] = down[x+1][y],down[x][y]
        neighboors.append(down)

    # Left
    if y-1>=0:
        left = copy.deepcopy(main)
        left[x][y],left[x][y-1] = left[x][y-1],left[x][y]
        neighboors.append(left)
    return neighboors


def dfs(start,goal):
    visited = []

    # The frontier is a stack
    # frontier = [ (state, path) ]
    
    fronter = [(start,[start])]
    while len(fronter)>0:
        curr_state,curr_path = fronter.pop(-1)
        if curr_state==goal:
            return curr_path
        if curr_state not in visited:
            visited.append(curr_state)
            # Get curr neighboors
            neighboors = get_neighboors(curr_state)
            for neighboor_state in neighboors:
                fronter.append((neighboor_state,curr_path+[neighboor_state]))
    return "No Solution found!"
```

Display State <b>(Optional)</b>

```python
def display_state(state):
    for row in range(len(state)):
        for col in range(len(state)):
            if state[row][col]!=0:
                print(state[row][col],end='\t')
            else:
                print(' ',end='\t')
        print()

def display_solution(solution):
    # Display actions
    for box in solution:
            display_state(box)
            print("\n")
```

Executions

```python
easy = [
        [1, 2, 3],
        [8, 0, 4],
        [7, 6, 5]]

solution = dfs(easy,goal)
print(f"EASY Soultion: {len(solution)} Steps:")
display_solution(solution)
```

```python
med = [
        [2, 8, 1],
        [0, 4, 3],
        [7, 6, 5]]

solution = dfs(med,goal)
print(f"Medium Soultion: {len(solution)} Steps\n")
display_solution(solution)
```

```python
hard = [
        [2, 8, 1],
        [4, 6, 3],
        [0, 7, 5]]

solution = dfs(hard,goal)
print(f"Hard Soultion: {len(solution)} Steps\n")
display_solution(solution)
```

```python
worst = [
        [5, 6, 7],
        [4, 0, 8],
        [3, 2, 1]]
solution = dfs(worst,goal)
print(f"Worst Soultion: {len(solution)} Steps\n")
display_solution(solution)
```
