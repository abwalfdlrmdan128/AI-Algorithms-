```python
import heapq
```

```python
graph = {
    "Arad":[("Timisoara",118),("Sibiu",140),("Zerind",75)],
    "Zerind":[("Arad",75),("Oradea",71)],
    "Oradea":[("Zerind",71),("Sibiu",151)],
    "Sibiu":[("Arad",140),("Rimnicu Vilcea",80),("Fagaras",99),("Oradea",151)],
    "Fagaras":[("Sibiu",99),("Bucharest",211)],
    "Rimnicu Vilcea":[("Sibiu",80),("Pitesti",97),("Craiova",146)],
    "Pitesti":[("Rimnicu Vilcea",97),("Bucharest",101),("Craiova",138)],
    "Craiova":[("Pitesti",138),("Rimnicu Vilcea",146),("Dobreta",120)],
    "Dobreta":[("Craiova",120),("Mehadia",75)],
    "Mehadia":[("Dobreta",75),("Lugoj",70)],
    "Lugoj":[("Mehadia",70),("Timisoara",111)],
    "Timisoara":[("Lugoj",111),("Arad",118)],
    "Bucharest":[("Fagaras",211),("Pitesti",101),("Giurgiu",90),("Urziceni",85)],
    "Giurgiu":[("Bucharest",90)],
    "Urziceni":[("Bucharest",85),("Hirsova",98),("Vaslui",142)],
    "Hirsova":[("Urziceni",98),("Eforie",86)],
    "Eforie":[("Hirsova",86)],
    "Vaslui":[("Urziceni",142),("Iasi",92)],
    "Iasi":[("Vaslui",92),("Neamt",87)],
    "Neamt":[("Iasi",87)]
}

heuristics = {
    "Arad": 366,
    "Bucharest": 0,
    "Craiova": 160,
    "Dobreta": 242,
    "Eforie": 161,
    "Fagaras": 176,
    "Giurgiu": 77,
    "Hirsova": 151,
    "Iasi": 226,
    "Lugoj": 244,
    "Mehadia": 241,
    "Neamt": 234,
    "Oradea": 380,
    "Pitesti": 100,
    "Rimnicu Vilcea": 193,
    "Sibiu": 253,
    "Timisoara": 329,
    "Urziceni": 80,
    "Vaslui": 199,
    "Zerind": 374
}

```

Solving Romania Problem using <b>Greedy Best First Search</b>

```python
def greedy_best_first(start, goal):
    visited = []
    # The frontier is a priorty queue
    # fronter = [ ( heuristic cost, city, path ) ] 
    fronter = [ (heuristics[start] , start, [start]) ]
    while len(fronter)>0:
        # curr = ( heuristic cost, city, path )
        curr_cost, curr_city, curr_path = heapq.heappop(fronter)
        if curr_city==goal:
            return curr_path
        if curr_city not in visited:
            visited.append(curr_city)
            # Get curr neighboors
            neighboors = graph[curr_city]
            for city,cost in neighboors:
                heapq.heappush(fronter, ( heuristics[city], city, curr_path+[city] ))
    return "No Solution found!"
```

Calculate Cost <b>(Optional)</b>

```python
# Calculate the path cost
def cal_cost(sol):
    sum = 0
    for i in range(len(sol)-1):
        ns = graph[sol[i]]
        for j in ns:
            if j[0]==sol[i+1]:
                sum+=j[1]
                break
    return sum
```

Execution

```python
sol = greedy_best_first("Arad","Bucharest")
print(f"{sol}\nTotal Cost = {cal_cost(sol)}")
```
